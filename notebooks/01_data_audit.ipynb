# 01. Data Audit: State Electricity Analysis

This notebook performs an exhaustive inspection and quality audit across all four project datasets in `datasets/`.


```python
import sys
sys.path.append('../src')
from dvp_project.data_loader import load_all_datasets, audit_datasets
import pandas as pd
```

## 1. Load All Raw Datasets


```python
datasets = load_all_datasets()
for name, df in datasets.items():
    print(f"{name}: {df.shape[0]} rows, {df.shape[1]} columns")
```

## 2. Comprehensive Quality Audit


```python
audit = audit_datasets(datasets)
audit_rows = []
for k, v in audit.items():
    cov = f"{v['date_coverage'][0]} to {v['date_coverage'][1]}" if v['date_coverage'] else 'N/A'
    audit_rows.append({
        'Dataset': k,
        'Rows': v['rows'],
        'Columns': v['columns'],
        'Missing Cells': v['missing_total'],
        'Duplicates': v['duplicates'],
        'Date Coverage': cov,
    })
pd.DataFrame(audit_rows)
```

## 3. Detailed Missing Values Breakdown


```python
for name, info in audit.items():
    print(f'=== {name} Missing Values ===')
    print(info['missing_values'] if info['missing_values'] else 'Zero missing values')
```
